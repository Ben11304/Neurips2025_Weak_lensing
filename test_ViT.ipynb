```python
import torch
import torch.nn as nn
import torch.optim as optim
from vit_pytorch import ViT
from torch.utils.data import DataLoader, TensorDataset
import numpy as np

# Set random seed for reproducibility
torch.manual_seed(42)

# Initialize the ViT model
v = ViT(
    image_size=256,
    patch_size=32,
    num_classes=4,
    dim=1024,
    depth=6,
    heads=16,
    mlp_dim=2048,
    dropout=0.1,
    emb_dropout=0.1
)

# Move model to GPU if available
device = torch.device("mps" if torch.mps.is_available() else "cpu")
v = v.to(device)

# Create a synthetic dataset (replace with your actual dataset)
num_samples = 1000
images = torch.randn(num_samples, 256, 256)  # Random images
labels = torch.randint(0, 4, (num_samples,)) 
print(labels.size(), images.size())    # Random labels (0-3 for 4 classes)
dataset = TensorDataset(images, labels)
dataloader = DataLoader(dataset, batch_size=128, shuffle=True)

# Define loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(v.parameters(), lr=0.001)

# Training loop
num_epochs = 10
v.train()
for epoch in range(num_epochs):
    running_loss = 0.0
    for i, (inputs, targets) in enumerate(dataloader):
        inputs, targets = inputs.to(device), targets.to(device)
        
        # Zero the parameter gradients
        optimizer.zero_grad()
        
        # Forward pass
        outputs = v(inputs)
        loss = criterion(outputs, targets)
        
        # Backward pass and optimize
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        
        if i % 10 == 9:  # Print every 10 batches
            print(f"[Epoch {epoch + 1}, Batch {i + 1}] Loss: {running_loss / 10:.4f}")
            running_loss = 0.0

    print(f"Epoch {epoch + 1} completed")

print("Training finished!")
```
